```python
import numpy as np

m = 4  # dimension of each vector
# all nonzero binary vectors of length m
D = np.array(
    [[int(x) for x in format(i, f"0{m}b")] for i in range(1, 2**m)]
).T  # shape (m, 2**m - 1)

# normalize columns
D = D / np.linalg.norm(D, axis=0, keepdims=True)

print(D)

# compute mutual coherence
G = D.T @ D
np.fill_diagonal(G, 0)
mu = np.max(np.abs(G))
spark_lb = 1 + 1 / mu

print(f"D shape: {D.shape}")
print(f"Mutual coherence: {mu:.3f}")
print(f"Spark lower bound: {spark_lb:.2f}")
```
